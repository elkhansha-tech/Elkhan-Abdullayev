```python
# Task 1: Introduction

# Create variables
name = "Elkhan"
age = 19
height = 1.78
is_student = True

# Print information using f-string
print(f"My name is {name}. I am {age} years old. My height is {height} meters. Student: {is_student}")

# Check data types
print(type(name))
print(type(age))
print(type(height))
print(type(is_student))

# Task 2: Calculator

# Get two numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Mathematical operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2
floor_division = num1 // num2
remainder = num1 % num2
power = num1 ** num2

# Display results
print(f"Addition: {addition}")
print(f"Subtraction: {subtraction}")
print(f"Multiplication: {multiplication}")
print(f"Division: {division:.2f}")
print(f"Floor division: {floor_division}")
print(f"Remainder: {remainder}")
print(f"Power: {power}")

# Task 3: Number Properties

number = int(input("Enter a positive integer: "))
last_digit = number % 10
is_even = number % 2 == 0
divisible_by_5 = number % 5 == 0
print(f"Last digit: {last_digit}")
print(f"Is even: {is_even}")
print(f"Divisible by 5: {divisible_by_5}") 

# Task 4: Three-digit Number

number = int(input("Enter a three-digit number: "))
hundreds = number // 100
tens = (number // 10) % 10
ones = number % 10
hundreds = number // 100
tens = (number // 10) % 10
ones = number % 10
digit_sum = hundreds + tens + ones
reversed_number = ones * 100 + tens * 10 + hundreds
print(f"Sum of digits: {digit_sum}")
print(f"Reversed number: {reversed_number}")

# Task 5: Leap Year

year = int(input("Enter a year: "))

is_leap_year = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

print(f"Is leap year: {is_leap_year}")

# Task 6: Autobuses

students = int(input("Enter number of students: "))

buses = (students + 39) // 40
empty_seats = buses * 40 - students

print(f"Buses needed: {buses}")
print(f"Empty seats: {empty_seats}")
```
